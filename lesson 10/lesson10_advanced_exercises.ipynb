## Exercise 1: Array statistics

- Create numeric array/list with at least 12 elements.
- Compute min, max, sum, average, and median (manual or built-in helper).
- Print all metrics.

```python
values = [12, 4, 9, 15, 7, 20, 3, 11, 18, 6, 14, 8]
sorted_values = sorted(values)
mid = len(sorted_values) // 2
if len(sorted_values) % 2:
    median = sorted_values[mid]
else:
    median = (sorted_values[mid - 1] + sorted_values[mid]) / 2
print("min:", min(values))
print("max:", max(values))
print("sum:", sum(values))
print("average:", sum(values) / len(values))
print("median:", median)
```

## Exercise 2: Iterator protocol deep dive

- Create iterator from a string and consume values with `next()` until exhausted.
- Handle `StopIteration` explicitly.
- Print total number of consumed characters.

```python
text = "Python"
char_iterator = iter(text)
consumed = 0
while True:
    try:
        character = next(char_iterator)
    except StopIteration:
        break
    print(character)
    consumed += 1
print("Total consumed:", consumed)
```

## Exercise 3: Custom iterator class

- Implement class `EvenNumbers(limit)` yielding even numbers up to limit.
- Use `__iter__` and `__next__`.
- Iterate and print values.

```python
class EvenNumbers:
    def __init__(self, limit):
        self.limit = limit
        self.current = 0

    def __iter__(self):
        return self

    def __next__(self):
        self.current += 2
        if self.current > self.limit:
            raise StopIteration
        return self.current

for number in EvenNumbers(20):
    print(number)
```

## Exercise 4: Iterator state with `next()`

- Create a custom iterator class `Countdown(start)` that counts down to 1.
- Use `iter()` and `next()` to take the first two values manually.
- Then use a `for` loop on the same iterator to get the remaining values.
- Add a short comment explaining why the `for` loop continues where `next()` stopped.

```python
class Countdown:
    def __init__(self, start):
        self.current = start

    def __iter__(self):
        return self

    def __next__(self):
        if self.current < 1:
            raise StopIteration
        value = self.current
        self.current -= 1
        return value

countdown = iter(Countdown(5))
print(next(countdown))
print(next(countdown))
for number in countdown:
    print(number)
# The for loop keeps using this same iterator, whose state advanced twice.
```
