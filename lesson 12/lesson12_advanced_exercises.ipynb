## Exercise 1: Robust numeric parser

- Write a function that parses user-provided string to int or float.
- Raise/handle exceptions for invalid formats.
- Return tuple `(success, value_or_error)` and print examples.

```python
def parse_number(text):
    """Parse a string as an integer first, then as a float."""
    try:
        return True, int(text)
    except ValueError:
        try:
            return True, float(text)
        except ValueError as error:
            return False, f"Invalid number: {error}"

for sample in ("42", "3.14", "hello"):
    print(sample, "->", parse_number(sample))
```

## Exercise 2: Multi-exception workflow

- Build a division function that converts both inputs with `float()` and divides, using `try/except/else/finally`.
- Handle `TypeError`, `ValueError`, and `ZeroDivisionError` separately.
- Demonstrate all branches with different calls.

```python
def safe_divide(first, second):
    try:
        numerator = float(first)
        denominator = float(second)
        result = numerator / denominator
    except TypeError as error:
        print("Type error:", error)
    except ValueError as error:
        print("Invalid number:", error)
    except ZeroDivisionError:
        print("Cannot divide by zero.")
    else:
        print("Result:", result)
    finally:
        print("Division attempt finished")

safe_divide("12", "3")
safe_divide("twelve", "3")
safe_divide("12", "0")
safe_divide(None, "3")
```

## Exercise 3: None-aware processing

- Given list with values and `None`, compute average of valid numbers only.
- Ignore `None` safely.
- Print cleaned list and average.

```python
values = [10, None, 20, None, 30]
cleaned = [value for value in values if value is not None]
if cleaned:
    average = sum(cleaned) / len(cleaned)
    print("Cleaned values:", cleaned)
    print("Average:", average)
else:
    print("No numeric values to average")
```

## Exercise 4: Input validation loop

- Prompt repeatedly for integer between 1 and 100 until valid.
- Use exception handling and custom error messages.
- Print accepted value and attempt count.

```python
attempts = 0
while True:
    attempts += 1
    try:
        value = int(input("Enter an integer from 1 to 100: "))
        if not 1 <= value <= 100:
            print("Out of range. Choose a number from 1 to 100.")
            continue
    except ValueError:
        print("That was not a whole number. Try again.")
        continue
    print(f"Accepted {value} after {attempts} attempt(s).")
    break
```

## Exercise 5: Mini CLI app

- Create text menu app (add/subtract/multiply/divide/quit).
- Use input + try/except for invalid choices and values.
- Keep looping until quit command.

```python
while True:
    print("\\nChoose: add, subtract, multiply, divide, quit")
    choice = input("Operation: ").strip().lower()
    if choice == "quit":
        print("Goodbye!")
        break
    if choice not in {"add", "subtract", "multiply", "divide"}:
        print("Unknown choice. Please choose a listed operation.")
        continue

    try:
        first = float(input("First number: "))
        second = float(input("Second number: "))
        if choice == "add":
            result = first + second
        elif choice == "subtract":
            result = first - second
        elif choice == "multiply":
            result = first * second
        else:
            result = first / second
    except ValueError:
        print("Please enter valid numbers.")
    except ZeroDivisionError:
        print("Cannot divide by zero.")
    else:
        print("Result:", result)
```
